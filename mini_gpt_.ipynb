```python


import urllib.request

url = "https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt"

data = urllib.request.urlopen(url).read().decode("utf-8")

print("Dataset downloaded successfully!")

print("Total characters:", len(data))

print("\nFirst 500 characters of our dataset:\n")

print(data[:500])
```

```python
chars = sorted(list(set(data)))
vocab_size = len(chars)
print("Unique characters:", ''.join(chars))
print("Vocabulary size:", vocab_size)
```

```python

stoi = {ch: i for i, ch in enumerate(chars)}
itos = {i: ch for i, ch in enumerate(chars)}

print("Character to Number:")
print(stoi)

print("\nNumber to Character:")
print(itos)
```

```python
sample = "Hello"
encoded = [stoi[c] for c in sample]
print("Original text:", sample)
print("Encoded numbers:", encoded)
```

```python

encoded_data = [stoi[c] for c in data]
print("First 100 encoded numbers:")
print(encoded_data[:100])
print("\nTotal encoded characters:", len(encoded_data))
```

```python
import torch
data_tensor = torch.tensor(encoded_data, dtype=torch.long)
split = int(0.9 * len(data_tensor))
train_data = data_tensor[:split]
val_data = data_tensor[split:]

print("Total data:", len(data_tensor))
print("Training data:", len(train_data))
print("Testing data:", len(val_data))
```

```python

block_size = 64
start = torch.randint(
    len(train_data) - block_size,
    (1,)
).item()
x = train_data[start:start + block_size]
y = train_data[start + 1:start + block_size + 1]

print("Input sequence:")
print(x)

print("\nTarget sequence:")
print(y)

print("\nInput shape:", x.shape)
print("Target shape:", y.shape)
```

```python
batch_size = 32
block_size = 64

def get_batch():
    ix = torch.randint(len(train_data) - block_size, (batch_size,))

    x = torch.stack([train_data[i:i+block_size] for i in ix])
    y = torch.stack([train_data[i+1:i+block_size+1] for i in ix])

    return x, y

xb, yb = get_batch()

print("Input batch shape:", xb.shape)
print("Target batch shape:", yb.shape)
```

```python
import torch.nn as nn
vocab_size = len(chars)
block_size = 64
n_embd = 128
token_embedding_table = nn.Embedding(vocab_size, n_embd)
position_embedding_table = nn.Embedding(block_size, n_embd)
tok_emb = token_embedding_table(xb)

pos = torch.arange(block_size)
pos_emb = position_embedding_table(pos)

x = tok_emb + pos_emb

print("Token embedding shape:", tok_emb.shape)
print("Position embedding shape:", pos_emb.shape)
print("Final embedding shape:", x.shape)
```

```python
head_size = 32
key = nn.Linear(n_embd, head_size, bias=False)
query = nn.Linear(n_embd, head_size, bias=False)
value = nn.Linear(n_embd, head_size, bias=False)
k = key(x)
q = query(x)
v = value(x)

print("Query shape:", q.shape)
print("Key shape:", k.shape)
print("Value shape:", v.shape)
```

```python
import torch.nn.functional as F

wei = q @ k.transpose(-2, -1)

wei = wei * (head_size ** -0.5)
tril = torch.tril(torch.ones(block_size, block_size))

wei = wei.masked_fill(tril == 0, float('-inf'))
wei = F.softmax(wei, dim=-1)

print("Attention scores shape:", wei.shape)
print(wei[0])
```

```python

out = wei @ v

print("Output shape:", out.shape)
print(out)
```

```python

feed_forward = nn.Sequential(
    nn.Linear(n_embd, 4 * n_embd),
    nn.ReLU(),
    nn.Linear(4 * n_embd, n_embd)
)

ff_out = feed_forward(tok_emb + pos_emb)

print("Feed Forward output shape:", ff_out.shape)
```

```python
class TransformerBlock(nn.Module):
    def __init__(self, n_embd, n_heads, block_size):
        super().__init__()

        self.attention = nn.MultiheadAttention(
            embed_dim=n_embd,
            num_heads=n_heads,
            batch_first=True
        )

        self.ln1 = nn.LayerNorm(n_embd)
        self.ln2 = nn.LayerNorm(n_embd)

        self.feed_forward = nn.Sequential(
            nn.Linear(n_embd, 4 * n_embd),
            nn.ReLU(),
            nn.Linear(4 * n_embd, n_embd)
        )


        mask = torch.triu(
            torch.ones(block_size, block_size, dtype=torch.bool),
            diagonal=1
        )

        self.register_buffer("mask", mask)

    def forward(self, x):
        attn_out, _ = self.attention(
            x, x, x,
            attn_mask=self.mask,
            need_weights=False
        )

        x = x + self.ln1(attn_out)
        x = x + self.ln2(self.feed_forward(x))

        return x


block = TransformerBlock(n_embd=128, n_heads=4, block_size=64)

block_out = block(tok_emb + pos_emb)

print("Transformer output shape:", block_out.shape)
```

```python
class MiniGPT(nn.Module):
    def __init__(self):
        super().__init__()

        self.token_embedding = nn.Embedding(vocab_size, n_embd)
        self.position_embedding = nn.Embedding(block_size, n_embd)

        self.transformer = TransformerBlock(
            n_embd=128,
            n_heads=4,
            block_size=64
        )

        self.final_norm = nn.LayerNorm(n_embd)

        self.output_layer = nn.Linear(n_embd, vocab_size)

    def forward(self, idx, targets=None):

        B, T = idx.shape

        # Token and position embeddings
        tok_emb = self.token_embedding(idx)
        pos = torch.arange(T, device=idx.device)
        pos_emb = self.position_embedding(pos)

        x = tok_emb + pos_emb

        # Transformer
        x = self.transformer(x)

        # Predict next character
        x = self.final_norm(x)
        logits = self.output_layer(x)

        loss = None

        if targets is not None:
            B, T, C = logits.shape
            logits = logits.view(B * T, C)
            targets = targets.view(B * T)

            loss = F.cross_entropy(logits, targets)

        return logits, loss


# Create our GPT model
model = MiniGPT()

# Test the model
logits, loss = model(xb, yb)

print("Model created successfully!")
print("Output shape:", logits.shape)
print("Initial loss:", loss.item())
```

```python
optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.001
)

print("Optimizer created successfully!")
```

```python
batch_size = 32
max_iters = 3000

for step in range(max_iters):

    # Get training data
    xb, yb = get_batch()

    # Forward pass
    logits, loss = model(xb, yb)

    # Reset gradients
    optimizer.zero_grad(set_to_none=True)

    # Calculate gradients
    loss.backward()

    # Update model parameters
    optimizer.step()

    # Display progress
    if step % 300 == 0:
        print(f"Step: {step}, Loss: {loss.item():.4f}")

print("Training completed!")
```

```python
@torch.no_grad()
def generate(idx, max_new_tokens):

    for _ in range(max_new_tokens):

        # Take the last 64 characters
        idx_cond = idx[:, -block_size:]

        # Get predictions
        logits, loss = model(idx_cond)

        # Get predictions for the last character
        logits = logits[:, -1, :]

        # Convert into probabilities
        probs = F.softmax(logits, dim=-1)

        # Select the next character
        idx_next = torch.multinomial(probs, num_samples=1)

        # Add the new character
        idx = torch.cat((idx, idx_next), dim=1)

    return idx


# Generate text
context = torch.zeros((1, 1), dtype=torch.long)

generated = generate(context, max_new_tokens=500)

# Convert numbers back into characters
generated_text = ''.join(
    itos[int(i)] for i in generated[0]
)

print(generated_text)
```

```python
import types

def fixed_forward(self, x):
    T = x.size(1)

    attn_out, _ = self.attention(
        x, x, x,
        attn_mask=self.mask[:T, :T],
        need_weights=False
    )

    x = x + self.ln1(attn_out)
    x = x + self.ln2(self.feed_forward(x))

    return x

model.transformer.forward = types.MethodType(
    fixed_forward, model.transformer
)

print("Attention mask fixed successfully!")
```

```python
max_iters = 5000

for step in range(max_iters):

    xb, yb = get_batch()

    logits, loss = model(xb, yb)

    optimizer.zero_grad(set_to_none=True)

    loss.backward()

    optimizer.step()

    if step % 500 == 0:
        print(f"Step: {step}, Loss: {loss.item():.4f}")

print("Additional training completed!")
```

```python
import gradio as gr
import torch
import torch.nn.functional as F

def generate_text(prompt, max_tokens):

    # Convert user input into numbers
    if prompt.strip() == "":
        context = torch.zeros((1, 1), dtype=torch.long)
    else:
        encoded = [stoi[ch] for ch in prompt if ch in stoi]
        if not encoded:
            return "Please enter characters available in the training vocabulary."
        context = torch.tensor([encoded], dtype=torch.long)

    # Generate text
    generated = generate(context, max_new_tokens=int(max_tokens))

    # Convert numbers back into characters
    result = ''.join(itos[int(i)] for i in generated[0])

    return result


# Create website
with gr.Blocks(theme=gr.themes.Soft()) as demo:

    gr.Markdown("# 🎭 Mini GPT – Shakespeare AI")
    gr.Markdown("### Generate Shakespeare-style text using a GPT model built from scratch!")

    prompt = gr.Textbox(
        label="Enter your starting text",
        placeholder="Example: ROMEO:",
        lines=3
    )

    max_tokens = gr.Slider(
        minimum=100,
        maximum=1000,
        value=500,
        step=100,
        label="Text length"
    )

    generate_button = gr.Button("✨ Generate Text")

    output = gr.Textbox(
        label="AI Generated Text",
        lines=15
    )

    generate_button.click(
        fn=generate_text,
        inputs=[prompt, max_tokens],
        outputs=output
    )

demo.launch(share=True)
```
